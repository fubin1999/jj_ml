```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
mono_data = pd.read_csv('../data/mono.csv')
multi_data  = pd.read_csv('../data/multi.csv')
```

```python
def clean_data(data):
    to_drop = ['Number', 'Sample number', 'Sample collection date', 'Name', 'Birth date', 'First PD date', 'End date', 'Height', 'Weight', 'CVD category']
    data = data.drop(to_drop, axis=1)
    return(data)

clean_mono_data = clean_data(mono_data)
clean_multi_data = clean_data(multi_data)
```

```python
train_data, test_data = train_test_split(clean_mono_data, test_size=0.3, stratify=clean_mono_data['Endpoint'], random_state=42)
```

```python
train_data.to_csv('../data/train.csv', index=False)
test_data.to_csv('../data/test1.csv', index=False)
clean_multi_data.to_csv('../data/test2.csv', index=False)
```
