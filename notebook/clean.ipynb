```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
mono_data = pd.read_csv('../data/mono.csv')
multi_data  = pd.read_csv('../data/multi.csv')
```

```python
def clean_data(data):
    to_drop = ['Number', 'Sample collection date', 'Name', 'Birth date', 'First PD date', 'End date', 'Height', 'Weight', 'CVD category']
    data = data.drop(to_drop, axis=1)
    return(data)

clean_mono_data = clean_data(mono_data)
clean_multi_data = clean_data(multi_data)
```

```python
common_columns = list(set(clean_mono_data.columns) & set(clean_multi_data.columns))
clean_mono_data = clean_mono_data[common_columns]
clean_multi_data = clean_multi_data[common_columns]
```

```python
train_data, test_data = train_test_split(clean_mono_data, test_size=0.3, stratify=clean_mono_data['Endpoint'], random_state=42)
```

```python
train_data = train_data.set_index("Sample number")
test_data = test_data.set_index("Sample number")
clean_multi_data = clean_multi_data.set_index("Sample number")
```

```python
train_data.to_csv('../data/train.csv')
test_data.to_csv('../data/test1.csv')
clean_multi_data.to_csv('../data/test2.csv')
```
