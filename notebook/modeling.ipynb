```python
import pandas as pd
import numpy as np

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegressionCV
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn import set_config

set_config(transform_output="pandas")
```

```python
train_data = pd.read_csv("../data/train.csv", index_col=0)
X_train = train_data.drop(columns=["Endpoint"])
y_train = train_data["Endpoint"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

keep_features = ['A2MG', 'APOB']
select_features = [col for col in X_train.columns if col not in keep_features]

model_pep = Pipeline([
    ('impute', SimpleImputer(strategy="mean")),
    ('scaler', StandardScaler()),
    ('fs', ColumnTransformer([
        ("passthrough", "passthrough", keep_features)
    ])),
    ('logreg', LogisticRegressionCV(class_weight="balanced")),
])

model_cli = Pipeline([
    ('impute', SimpleImputer(strategy="mean")),
    ('scaler', StandardScaler()),
    ('fs', ColumnTransformer([
        ("kbest", SelectKBest(f_classif, k=6), select_features),
    ])),
    ('logreg', LogisticRegressionCV(class_weight="balanced")),
])

model_all = Pipeline([
    ('impute', SimpleImputer(strategy="mean")),
    ('scaler', StandardScaler()),
    ('fs', ColumnTransformer([
        ("kbest", SelectKBest(f_classif, k=6), select_features),
        ("passthrough", "passthrough", keep_features)
    ])),
    ('logreg', LogisticRegressionCV(class_weight="balanced")),
])

model_pep.fit(X_train, y_train)
model_cli.fit(X_train, y_train)
model_all.fit(X_train, y_train)
```

```python
test_data1 = pd.read_csv("../data/test1.csv", index_col=0)
test_data2 = pd.read_csv("../data/test2.csv", index_col=0)

X_test1 = test_data1[X_train.columns]
y_test1 = test_data1["Endpoint"]

y_pred1_pep = model_pep.predict(X_test1)
y_proba1_pep = model_pep.predict_proba(X_test1)[:, 1]
y_pred1_cli = model_cli.predict(X_test1)
y_proba1_cli = model_cli.predict_proba(X_test1)[:, 1]
y_pred1_all = model_all.predict(X_test1)
y_proba1_all = model_all.predict_proba(X_test1)[:, 1]

X_test2 = test_data2[X_train.columns]
y_test2 = test_data2["Endpoint"]

y_pred2_pep = model_pep.predict(X_test2)
y_proba2_pep = model_pep.predict_proba(X_test2)[:, 1]
y_pred2_cli = model_cli.predict(X_test2)
y_proba2_cli = model_cli.predict_proba(X_test2)[:, 1]
y_pred2_all = model_all.predict(X_test2)
y_proba2_all = model_all.predict_proba(X_test2)[:, 1]
```

```python
result1_pep = pd.DataFrame({
    "sample_number": y_test1.index,
    "feature_type": "Only Peptides",
    "true": y_test1,
    "pred": y_pred1_pep,
    "prob": y_proba1_pep
})

result1_cli = pd.DataFrame({
    "sample_number": y_test1.index,
    "feature_type": "Only Clinics",
    "true": y_test1,
    "pred": y_pred1_cli,
    "prob": y_proba1_cli
})

result1_all = pd.DataFrame({
    "sample_number": y_test1.index,
    "feature_type": "All Features",
    "true": y_test1,
    "pred": y_pred1_all,
    "prob": y_proba1_all
})

# merge result
result1 = pd.concat([result1_pep, result1_cli, result1_all])

result2_pep = pd.DataFrame({
    "sample_number": y_test2.index,
    "feature_type": "Only Peptides",
    "true": y_test2,
    "pred": y_pred2_pep,
    "prob": y_proba2_pep
})

result2_cli = pd.DataFrame({
    "sample_number": y_test2.index,
    "feature_type": "Only Clinics",
    "true": y_test2,
    "pred": y_pred2_cli,
    "prob": y_proba2_cli
})

result2_all = pd.DataFrame({
    "sample_number": y_test2.index,
    "feature_type": "All Features",
    "true": y_test2,
    "pred": y_pred2_all,
    "prob": y_proba2_all
})

# merge result
result2 = pd.concat([result2_pep, result2_cli, result2_all])
```

```python
result1.to_csv("../results/data/result_test1.csv", index=False)
result2.to_csv("../results/data/result_test2.csv", index=False)
```
